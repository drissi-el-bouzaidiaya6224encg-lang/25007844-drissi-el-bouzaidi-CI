```python
# Sources réelles intégrées à scikit-learn : trois problèmes différents
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes, load_breast_cancer, load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, f1_score, roc_auc_score, silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
GRAINE = 42

jeu_regression = load_diabetes()
X_reg, y_reg = jeu_regression.data, jeu_regression.target
jeu_classification = load_breast_cancer()
X_cls, y_cls = jeu_classification.data, jeu_classification.target
jeu_non_supervise = load_wine()
X_clu = StandardScaler().fit_transform(jeu_non_supervise.data)  # Exploration sans cible

Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(X_reg, y_reg, test_size=.25, random_state=GRAINE)
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(X_cls, y_cls, test_size=.25, stratify=y_cls, random_state=GRAINE)
print('Diabetes:', X_reg.shape, '| Breast Cancer:', X_cls.shape, '| Wine:', X_clu.shape)
print('Ces données sont réelles ; les étiquettes du vin ne sont pas employées dans le clustering.')
```

## 1. Régression linéaire

**Variante C — question spécifique :** Quel diagnostic du résidu indique une mauvaise spécification linéaire ?

 Équation du modèle :$$\hat{y} = \beta_0 + \sum_{j=1}^p \beta_j X_j = \beta_0 + \beta_1 X_{t} + \beta_2 X_{t-1} + \beta_3 X_{t-2} + \beta_4 \cdot \text{taux\_croissance}$$

Définition des symboles :$\hat{y}$ : Valeur prédite à l'horizon $t+1$ (importations).$\beta_0$ : Terme d'interception (ordonnée à l'origine).$\beta_j$ : Coefficients de régression associés à chaque variable $X_j$.Principe de minimisation : Estimation des coefficients par les Moindres Carrés Ordinaires (MCO / OLS) en minimisant la somme des carrés des résidus ($RSS$) : $$\min_{\beta} \sum_{i=1}^n \left(y_i - \hat{y}_i\right)^2$$


Schéma original (Principe conceptuel)Schéma de flux : $$\left[ X_t, X_{t-1}, X_{t-2}, \text{taux} \right] \xrightarrow{\text{Combinaison linéaire } (\beta)} \hat{y} \xrightarrow{\text{Calcul de l'erreur } e_i = y_i - \hat{y}_i} \text{Minimisation de } \sum e_i^2$$Concept : Ajustement d'un hyperplan optimal à travers le nuage de points pour minimiser la somme des distances verticales au carré entre les valeurs réelles $y_i$ et les prédictions $\hat{y}_i$.



C. Contextes et choixJustification : Utilisée comme premier modèle paramétrique de référence (benchmark linéaire simple). Elle permet d'évaluer la relation directe entre l'historique de la demande et la valeur future.Limites dans notre contexte : Très sensible à la multicolinéarité entre les retards temporels ($X_t, X_{t-1}, X_{t-2}$) et au surapprentissage (overfitting) en raison de la faible taille de l'échantillon.



D. Test de sensibilité et interprétationHyperparamètre testé : fit_intercept=True vs fit_intercept=False.Résultat du code : $MAE = 41.55$ et $R^2 = 0.485$ (le modèle explique 48.5 % de la variance).Interprétation : Sans régularisation, les coefficients s'ajustent de manière non contrainte, ce qui peut amplifier le bruit en présence de colinéarité.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import LinearRegression
modele = LinearRegression().fit(Xr_ent, yr_ent)
predictions = modele.predict(Xr_test)
print('MAE régression linéaire :', round(mean_absolute_error(yr_test, predictions), 2))
print('R² :', round(r2_score(yr_test, predictions), 3))
```

## 2. Ridge

**Variante C — question spécifique :** Comment évoluent les coefficients quand alpha augmente ?

 1. Présentation (avec équation et symboles) :Équation de la fonction coût $L_1$ : $$\min_{\beta} \left[ \sum_{i=1}^n (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^p |\beta_j| \right]$$Définition des symboles : $y_i$ est la valeur réelle, $\hat{y}_i = \beta_0 + \sum \beta_j X_{j,i}$ est la prédiction, $\beta_j$ représente les coefficients du modèle, et $\alpha \ge 0$ est le paramètre de régularisation $L_1$.Propriété clé : La pénalité basée sur la somme des valeurs absolues ($\sum |\beta_j|$) possède des coins géométriques qui annulent exactement certains coefficients ($\beta_j = 0$), effectuant ainsi une sélection automatique des variables.
 2. Schéma original (Principe conceptuel) :Pipeline : $X \xrightarrow{\text{StandardScaler()}} X_{\text{centré-réduit}} \xrightarrow{\text{Lasso}(\alpha=2.0)} \hat{y}$Concept géométrique : La contrainte $L_1$ forme un losange/diamant dans l'espace des paramètres ($|\beta_1| + |\beta_2| \le t$). Les ellipses de niveau de la somme des carrés des erreurs rencontrent préférentiellement ce diamant sur ses sommets (sur les axes), ce qui met à zéro les variables redondantes ou peu contributives.
 3. Contextes et choix :Justification : Le modèle Lasso est privilégié lorsqu'on suspecte la présence de nombreuses variables non pertinentes ou colinéaires, afin d'obtenir un modèle plus simple et interprétable (parcimonieux).Rôle du Pipeline : L'utilisation de StandardScaler() est obligatoire car la pénalité $L_1$ dépend directement de l'échelle des variables. Sans normalisation préalable, les variables de grande magnitude seraient indûment préservées.
 4. Test de sensibilité et interprétation :Hyperparamètre modifié : $\alpha$ (intensité de la pénalité).Résultats observés :Pour $\alpha = 0.1$, la sévérité est faible (1 variable annulée).Pour $\alpha = 2.0$, la sélection est optimale (2 variables annulées), atteignant la meilleure erreur absolue moyenne ($\text{MAE} = 41.37$).Pour $\alpha = 10.0$, la pénalité est trop forte (6 variables annulées), ce qui élimine des informations utiles et dégrade la précision ($\text{MAE} = 43.64$, sous-apprentissage).
> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), Ridge(alpha=12.0)).fit(Xr_ent, yr_ent)
print('MAE Ridge :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

## 3. Lasso

**Variante C — question spécifique :** Combien de variables sont annulées pour trois valeurs d’alpha ?

1. Présentation (avec équation et symboles) :Équation du modèle polynomial de degré 2 avec Ridge : $$\hat{y} = \beta_0 + \sum_{j=1}^p \beta_j X_j + \sum_{j=1}^p \sum_{k=j}^p \beta_{j,k} X_j X_k$$ $$\min_{\beta} \left[ \sum_{i=1}^n (y_i - \hat{y}_i)^2 + \alpha \left( \sum \beta_j^2 + \sum \beta_{j,k}^2 \right) \right]$$Définition des symboles : $X_j$ sont les variables linéaires d'origine, $X_j X_k$ sont les termes d'interaction et les termes quadratiques ($X_j^2$), $\beta_{j,k}$ sont leurs coefficients respectifs, et $\alpha$ est la pénalité $L_2$ Ridge.
2. Schéma original (Principe conceptuel) :Pipeline : $X \xrightarrow{\text{StandardScaler()}} X_{\text{sc}} \xrightarrow{\text{PolynomialFeatures}(\text{degree}=2)} X_{\text{poly}} \xrightarrow{\text{Ridge}(\alpha=100.0)} \hat{y}$Concept : Projection des 10 variables initiales vers un espace enrichi de 65 caractéristiques non linéaires (incluant carrés et produits croisés), permettant d'ajuster des surfaces de décision courbes au lieu de simples hyperplans.
3. Contextes et choix :Justification : Utilisée pour capturer des effets non linéaires ou des interactions complexes entre les caractéristiques.Nécessité des hyperparamètres (include_bias=False et Ridge(alpha=100.0)) : La multiplication des variables crée une forte multicolinéarité et fait exploser la variance. L'ajout du régularisateur Ridge avec un $\alpha$ élevé est indispensable pour éviter l'effondrement des performances hors-échantillon.
4. Test de sensibilité et interprétation :Hyperparamètres modifiés : degree (1 vs 2) et alpha (0.1 vs 100.0).Résultats observés :Au degré 1 (linéaire, $\alpha=0.1$), $R^2_{\text{train}} = 0.519$ et $R^2_{\text{test}} = 0.485$.Au degré 2 sans régularisation suffisante ($\alpha=0.1$), le modèle surajuste sévèrement ($R^2_{\text{train}} = 0.603$, $R^2_{\text{test}} = 0.431$).Au degré 2 avec $\alpha=100.0$, la pénalité réfrène la complexité du modèle, permettant d'obtenir $R^2_{\text{test}} = 0.501$.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.linear_model import Lasso
modele = make_pipeline(StandardScaler(), Lasso(alpha=2.0, max_iter=15000)).fit(Xr_ent, yr_ent)
print('Nombre de coefficients non nuls :', np.count_nonzero(modele[-1].coef_))
print('MAE Lasso :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

## 4. Régression polynomiale

**Variante C — question spécifique :** Le degré 2 améliore-t-il le test ou seulement l’entraînement ?

1. Présentation (avec équation et symboles) :Équation de prédiction par partitionnement de l'espace : $$\hat{f}(x) = \sum_{m=1}^M c_m I(x \in R_m) \quad \text{où} \quad c_m = \frac{1}{N_m} \sum_{x_i \in R_m} y_i$$Définition des symboles : $R_m$ représente la région (feuille) $m$, $N_m$ est le nombre d'échantillons dans la feuille, $c_m$ est la moyenne des cibles $y_i$ dans cette région, et $I(\cdot)$ est la fonction indicatrice.Critère de séparation : Minimisation de l'impureté de variance à chaque nœud : $$\text{Critère} = \sum_{i \in R_{\text{gauche}}} (y_i - \bar{y}_{\text{gauche}})^2 + \sum_{i \in R_{\text{droite}}} (y_i - \bar{y}_{\text{droite}})^2$$2. Schéma original (Principe conceptuel) :Pipeline : $$X \xrightarrow{\text{Tests boursiers de décisions } (X_j \le t_k)} \text{Partition en hyperrectangles } R_m \xrightarrow{\text{Moyenne des } y} \hat{y}$$Concept géométrique : Découpage successif de l'espace des variables par des hyperplans orthogonaux aux axes des descripteurs pour isoler des sous-groupes homogènes de la cible $y$.
3. Contextes et choix :Justification : Modèle non paramétrique hautement interprétable (visualisable sous forme de structure décisionnelle en arbre). Il capture les non-linéarités sans nécessiter de transformation de variables.Choix des hyperparamètres (max_depth=3, min_samples_leaf=8) : Indispensables pour élaguer l'arbre et réduire sa variance naturelle élevée, qui est la principale faiblesse des arbres de décision individuels.4. Test de sensibilité et interprétation :Hyperparamètre modifié : max_depth (profondeur maximale).Résultats observés :Avec max_depth = 3, le modèle est équilibré (MAE Test = 44.57).En augmentant à max_depth = 6, la performance sur le jeu d'entraînement s'améliore (MAE Train passe de 44.66 à 35.23), mais la performance en test se dégrade (MAE Test grimpe à 49.37), illustrant l'augmentation de la variance liée à la profondeur.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele = make_pipeline(StandardScaler(), PolynomialFeatures(degree=2, include_bias=False), Ridge(alpha=100.)).fit(Xr_ent, yr_ent)
print('R² polynomial :', round(r2_score(yr_test, modele.predict(Xr_test)), 3))
```

## 6. Forêt aléatoire — régression

**Variante C — question spécifique :** La forêt réduit-elle la variance par rapport à l’arbre ?
1. Présentation (avec équation et symboles) :Équation de prédiction de la forêt aléatoire : $$\hat{f}_{\text{RF}}(x) = \frac{1}{B} \sum_{b=1}^B T_b(x; \Theta_b)$$Définition des symboles : $B$ est le nombre d'arbres (n_estimators), $T_b$ est le $b$-ième arbre de décision entraîné sur un échantillon bootstrap des données, et $\Theta_b$ représente les tirages aléatoires des variables d'entraînement à chaque séparation.
2. Schéma original (Principe conceptuel) :Pipeline d'Ensemble : $$X \xrightarrow{\text{B tirages Bootstrap}} \{X^{(1)}, \dots, X^{(B)}\} \xrightarrow{\text{Entraînement de B arbres décorrélés}} \{\hat{y}_1, \dots, \hat{y}_B\} \xrightarrow{\text{Moyenne des prédictions}} \hat{y} = \frac{1}{B}\sum \hat{y}_b$$Concept : L'agrégation de multiples arbres indépendants permet de lisser les erreurs individuelles et de réduire la variance globale d'un facteur proportionnel à la décorrélation des arbres.3. Contextes et choix :Justification : Algorithme non paramétrique très puissant et robuste. Il corrige l'instabilité des arbres de décision simples et offre une excellente capacité de généralisation sans nécessiter de normalisation des données.Choix des hyperparamètres (n_estimators=60, max_depth=6) : n_estimators=60 garantit la convergence de la moyenne de l'ensemble, tandis que max_depth=6 empêche la sur-complexité individuelle de chaque arbre.
4. Test de sensibilité et interprétation :Hyperparamètres modifiés : n_estimators (10, 60, 200) et max_depth (3, 6).Résultats observés :Avec n_estimators = 10 et max_depth = 6, MAE Test = 44.67.En augmentant à n_estimators = 60, MAE Test s'améliore à 43.11 (et 42.81 avec 200 arbres). L'augmentation du nombre d'arbres $B$ stabilise les prédictions et diminue la variance sans provoquer de surapprentissage supplémentaire.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.ensemble import RandomForestRegressor
modele = RandomForestRegressor(n_estimators=60, max_depth=6, n_jobs=-1, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE forêt :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

## 7. Gradient Boosting — régression

**Variante C — question spécifique :** Que change max_iter sur le coût et le risque de surajustement ?

1. Présentation (avec équation et symboles) :Équation de la combinaison additive séquentielle : $$\hat{f}_{M}(x) = \hat{f}_{0}(x) + \sum_{m=1}^{M} \nu \cdot h_m(x)$$Définition des symboles : $M$ est le nombre total d'itérations (max_iter), $h_m(x)$ est l'arbre de régression ajusté aux pseudo-résidus (gradient de la fonction de perte) à l'étape $m$, $\nu$ est le taux d'apprentissage (learning rate), et $\hat{f}_{0}(x)$ est l'initialisation constante (moyenne des cibles $y$).Descente de gradient fonctionnelle : À chaque itération $m$, l'arbre $h_m$ est entraîné pour approximer le gradient négatif : $$r_{i,m} = -\left[ \frac{\partial L(y_i, f(x_i))}{\partial f(x_i)} \right]_{f = \hat{f}_{m-1}}$$2. Schéma original (Principe conceptuel) :Pipeline séquentiel : $$X \xrightarrow{\text{Initialisation } \hat{f}_0} \text{Boucle } m=1 \dots M \xrightarrow{\text{Calcul des résidus/gradients } r_m} \text{Ajustement d'arbre } h_m \xrightarrow{\text{Mise à jour } \hat{f}_m = \hat{f}_{m-1} + \nu h_m} \hat{y}$$Concept : Construction d'un ensemble séquentiel d'arbres de décision peu profonds où chaque nouvel arbre vient corriger les erreurs de prédiction (résidus) accumulées par l'ensemble des arbres précédents.
3. Contextes et choix :Justification : Algorithme état de l'art pour les données tabulaires. L'implémentation HistGradientBoostingRegressor discrétise les variables numériques en histogrammes (bins), ce qui accélère considérablement l'entraînement sur de grands volumes et gère efficacement les valeurs manquantes.Choix des hyperparamètres (max_iter=100, max_leaf_nodes=9) : max_iter=100 contrôle le nombre maximum d'arbres construits, tandis que max_leaf_nodes=9 contraint la taille de chaque arbre (arbres peu profonds) pour prévenir le surapprentissage individuel.
4. Test de sensibilité et interprétation :Hyperparamètre modifié : max_iter (10, 50, 100, 500) avec max_leaf_nodes=9.Résultats observés :Pour max_iter = 10 : MAE Train = 46.24, MAE Test = 48.19 (sous-apprentissage).Pour max_iter = 50 : MAE Train = 28.68, MAE Test = 43.13 (compromis optimal).Pour max_iter = 100 (code de base) : MAE Train = 21.42, MAE Test = 44.18.Pour max_iter = 500 : MAE Train = 4.35, MAE Test = 47.29 (surapprentissage prononcé).
> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele = HistGradientBoostingRegressor(max_iter=100, max_leaf_nodes=9, random_state=GRAINE).fit(Xr_ent, yr_ent)
print('MAE boosting :', round(mean_absolute_error(yr_test, modele.predict(Xr_test)), 2))
```

## 8. Régression logistique

**Variante C — question spécifique :** Quel effet de C sur les probabilités et la frontière ?

1. Présentation (avec équation et symboles) :Équation de la probabilité a posteriori : $$P(Y=1 | X) = \sigma(z) = \frac{1}{1 + e^{-z}} \quad \text{où} \quad z = \beta_0 + \sum_{j=1}^p \beta_j X_j$$Fonction coût Log-Loss (avec pénalité $L_2$) : $$\min_{\beta} \left[ -\frac{1}{n}\sum_{i=1}^n \left( y_i \log(\hat{p}_i) + (1-y_i) \log(1-\hat{p}_i) \right) + \frac{1}{2C} \sum_{j=1}^p \beta_j^2 \right]$$Définition des symboles : $y_i \in \{0, 1\}$ est la classe binaire réelle, $\hat{p}_i = P(Y=1|X_i)$ est la probabilité prédite, $\beta_j$ sont les coefficients du modèle, et $C > 0$ est l'inverse du paramètre de régularisation $\lambda$ ($C = 1/\lambda$).
2. Schéma original (Principe conceptuel) :Pipeline : $$X \xrightarrow{\text{StandardScaler()}} X_{\text{sc}} \xrightarrow{\text{Combinaison linéaire } z = \beta^T X} \xrightarrow{\text{Fonction Sigmoïde } \sigma(z)} \hat{p} \in [0, 1] \xrightarrow{\text{Seuil } \ge 0.5} \hat{y} \in \{0, 1\}$$Concept géométrique : Séparation des deux classes par un hyperplan de décision linéaire dans l'espace des descripteurs scaliérisés, transformant la distance signée à la frontière en une probabilité continue entre 0 et 1 via la courbe sigmoïde.
3. Contextes et choix :Justification : Modèle de référence (benchmark) pour la classification binaire. Il fournit des prédictions probabilistes directement calibrées et des coefficients facilement interprétables en termes de ratios de chances (Odds Ratios : $e^{\beta_j}$).Choix du Pipeline et des hyperparamètres (StandardScaler(), C=0.5, max_iter=2000) :StandardScaler() est obligatoire car la régularisation $L_2$ dépend directement de l'échelle des descripteurs.C=0.5 introduit une régularisation modérée pour prévenir le surapprentissage sur les 30 variables du jeu de données.max_iter=2000 garantit la convergence complète de l'algorithme d'optimisation (L-BFGS).
4. Test de sensibilité et interprétation :Hyperparamètre modifié : $C$ (0.01, 0.1, 0.5, 1.0, 10.0, 100.0).Résultats observés :Pour $C = 0.01$ (forte pénalité) : F1 Train = 0.9674, F1 Test = 0.9519 (sous-ajustement léger).Pour $C = 0.5$ (code de base) : F1 Train = 0.9907, F1 Test = 0.9889 (équilibre optimal).Pour $C = 100.0$ (faible pénalité) : F1 Train = 0.9981, F1 Test = 0.9551 (dégradation du score F1 en test suite au surapprentissage de la frontière).
> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.

## 9. KNN

## 26. Epsilon-greedy

**Variante C — question spécifique :** Quel effet de epsilon sur la concentration des choix ?

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Exploration epsilon-greedy sur le même bandit simulé
rng = np.random.default_rng(GRAINE)
probabilites_simulees = np.array([.35, .55, .65, .45])
recompenses = np.zeros(4); nombre_tirages = np.zeros(4)
epsilon = .2
for etape in range(1500):
    bras = int(rng.integers(4)) if rng.random() < epsilon or not np.all(nombre_tirages) else int(np.argmax(recompenses / np.maximum(nombre_tirages, 1)))
    gain = float(rng.random() < probabilites_simulees[bras])
    recompenses[bras] += gain; nombre_tirages[bras] += 1
print('Fréquences de sélection :', nombre_tirages.astype(int))
print('Valeurs estimées :', np.round(recompenses / np.maximum(nombre_tirages, 1), 3))
```

## 27. Q-learning

**Variante C — question spécifique :** Étudiez les récompenses des épisodes, pas seulement Q moyen.

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Q-learning manuel — environnement CliffWalking (Gymnasium)
import gymnasium as gym
env = gym.make('CliffWalking-v0')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    for pas in range(120):
        action = int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
        suivant, recompense, termine, tronque, _ = env.step(action)
        cible = recompense + (0 if termine else .95 * np.max(q_table[suivant]))
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat = suivant
        if termine or tronque: break
print('Table Q :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

## 28. SARSA

**Variante C — question spécifique :** SARSA privilégie-t-il une trajectoire plus sûre sous exploration ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# SARSA manuel — même environnement, politique epsilon-greedy suivie
import gymnasium as gym
env = gym.make('CliffWalking-v0')
q_table = np.zeros((env.observation_space.n, env.action_space.n))
rng = np.random.default_rng(GRAINE)
def choisir_action(etat, epsilon):
    return int(rng.integers(env.action_space.n)) if rng.random() < epsilon else int(np.argmax(q_table[etat]))
for episode in range(1100):
    etat, _ = env.reset(seed=GRAINE + episode)
    epsilon = max(.08, .8 * (1 - episode / 900))
    action = choisir_action(etat, epsilon)
    for pas in range(120):
        suivant, recompense, termine, tronque, _ = env.step(action)
        action_suivante = choisir_action(suivant, epsilon)
        cible = recompense + (0 if termine else .95 * q_table[suivant, action_suivante])
        q_table[etat, action] += .15 * (cible - q_table[etat, action])
        etat, action = suivant, action_suivante
        if termine or tronque: break
print('Table Q SARSA :', q_table.shape, '| valeur moyenne :', round(q_table.mean(), 2))
env.close()
```

## 29. DQN — cible conceptuelle

**Variante C — question spécifique :** Pourquoi la cible est-elle égale à la récompense en état terminal ?

**À compléter :** Présentation (avec équation et symboles), Schéma original, Contextes et choix, Test de sensibilité et interprétation.

> Modifiez au moins un hyperparamètre et justifiez son effet ; conservez le résultat produit par le code.


```python
# Cibles DQN : démonstration du calcul de Bellman, SANS réseau neuronal
recompenses = np.array([-1., -100., 0.])
termine = np.array([False, True, False])
meilleure_valeur_suivante = np.array([-3., -2., 5.])
gamma = .9
cibles = recompenses + gamma * (~termine) * meilleure_valeur_suivante
print('Cibles de Bellman calculées :', np.round(cibles, 2))
print('Ce calcul ne constitue pas un entraînement DQN.')
```
